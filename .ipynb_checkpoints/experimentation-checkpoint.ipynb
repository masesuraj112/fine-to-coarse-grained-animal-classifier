```python
# Files from task1_data folder 

import pandas as pd 

# Given features from specifications 
additional_features_df = pd.read_csv('task1_data/additional_features.csv')
# Colour histograms
color_histogram_df = pd.read_csv('task1_data/color_histogram.csv')
# HOG features reduced by PCA
hog_pca_df = pd.read_csv('task1_data/hog_pca.csv')
# Unchanged Test data
test_metadata_df = pd.read_csv('task1_data/test_metadata.csv')
# Unchanged Training data 
train_metadata_df = pd.read_csv('task1_data/train_metadata.csv')
```

**Experiment 1.1: Basic SVM**

```python
# Naive SVM just to understand what to do

# In this example we will only merge with one additional feature just to get this working 

# In later iterations we will figure out how to improve it

# No Hyperparameter tuning 


from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score


# Associate features with correct labels for training purposes 
additional_features_merged = pd.merge(train_metadata_df, additional_features_df, on="image_id")


# stratify=stratify=train_metadata_df['class_name'] ensures no class imbalance when splitting the data 
# random_state=42 ensures that output is consistent per run
# Splitting the data into a training (which will be further split) and a validation set (which will not be touched)
# local_test_set refers to the test I will be testing for accuracy and other statistics 
training_temp, local_test_set = train_test_split(additional_features_merged, test_size=0.2, random_state=42, stratify=additional_features_merged['class_name'])

x_trained = training_temp.drop(columns=["image_id", "class_name", "image_path", "class_id"])
print(x_trained.columns)
y_trained = training_temp["class_name"]
svm_model = svm.SVC()
svm_model.fit(x_trained, y_trained)
x_test = local_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_test = local_test_set["class_name"]
predictions = svm_model.predict(x_test)
accuracy = accuracy_score(predictions, y_test)
print(accuracy)

```

**Experiment 1.2 : SVM with HyperParameter Testing**

```python
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV 
from sklearn.svm import SVC 




# Associate features with correct labels for training purposes 
additional_features_merged = pd.merge(train_metadata_df, hog_pca_df, on="image_id")

# stratify=stratify=train_metadata_df['class_name'] ensures no class imbalance when splitting the data 
# random_state=42 ensures that output is consistent per run
# Splitting the data into a training (which will be further split) and a validation set (which will not be touched)
# local_test_set refers to the test I will be testing for accuracy and other statistics 
training_temp, local_test_set = train_test_split(additional_features_merged, test_size=0.2, random_state=42, stratify=additional_features_merged['class_name'])

# Splitting training set further 
# 70% Training, 10% Validation, 20% Test
actual_training_set, actual_validation_set = train_test_split(training_temp, test_size=0.1, random_state=42, stratify=training_temp['class_name'])

# From https://www.geeksforgeeks.org/machine-learning/svm-hyperparameter-tuning-using-gridsearchcv-ml/
hyperparameters = {'C': [0.1, 1, 10, 100, 1000], 
			'gamma': [1, 0.1, 0.01, 0.001, 0.0001], 
			'kernel': ['rbf']}
# Hyperparameter tuning 
hyper_grid = GridSearchCV(SVC(), hyperparameters, refit = True) 
hyper_grid.fit(training_temp.iloc[:, 4:27], training_temp.iloc[:, 3])
predictions = hyper_grid.predict(local_test_set.iloc[:, 4:27])
accuracy = accuracy_score(predictions, local_test_set.iloc[:, 3])
print(accuracy)


# This attempt had an accuracy of 0.476 which is fairly close 
# Difference is I aggregrated with hog pca 




```

**Experiment 1.3 : SVM with PCA applied to additional_features**

```python
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

additional_features_merged = pd.merge(train_metadata_df, additional_features_df, on="image_id")

# Split
training_temp, local_test_set = train_test_split(
    additional_features_merged,
    test_size=0.2,
    random_state=42,
    stratify=additional_features_merged['class_name']
)

x_trained = training_temp.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_trained = training_temp["class_name"]

x_test = local_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_test = local_test_set["class_name"]

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('svc', svm.SVC())
])

hyperparameters = {
    'pca__n_components': [5, 10, 15, 20],
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': [1, 0.1, 0.01, 0.001],
    'svc__kernel': ['rbf']
}

# Grid search ONLY on training data
hyper_grid = GridSearchCV(pipeline, hyperparameters, refit=True)
hyper_grid.fit(x_trained, y_trained)

# Predict
predictions = hyper_grid.predict(x_test)

# Evaluate
accuracy = accuracy_score(y_test, predictions)
print(accuracy)
```

```python
# https://www.youtube.com/watch?v=7K-gnWlgAt8
# Extracting features from RESNET - doing own feature engineering 
import torch
from torchvision import models, transforms
import numpy
from PIL import Image
import os

model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
model.eval()


transform_pipeline = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )  
])

feature_extractor = torch.nn.Sequential(*list(model.children())[:-1])
BASE_DIR = os.getcwd() + "/task1_data/"

def extract_features(image_path):
    image = Image.open(image_path).convert("RGB")
    image = transform_pipeline(image)
    image = image.unsqueeze(0)
    with torch.no_grad():
        features = feature_extractor(image)
    features = features.detach().numpy().flatten()
    return features 

features_list = []
for idx, row in train_metadata_df.iterrows():
    image_path = os.path.join(BASE_DIR, row["image_path"])
    feats = extract_features(image_path)
    features_list.append(feats)

features_df = pd.DataFrame(features_list, index=train_metadata_df["image_id"])


```

```python
# Naive SVM just to understand what to do

# In this example we will only merge with one additional feature just to get this working 

# In later iterations we will figure out how to improve it

# No Hyperparameter tuning 


from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler


# Associate features with correct labels for training purposes 
additional_features_merged = pd.merge(train_metadata_df, features_df, on="image_id")


# stratify=stratify=train_metadata_df['class_name'] ensures no class imbalance when splitting the data 
# random_state=42 ensures that output is consistent per run
# Splitting the data into a training (which will be further split) and a validation set (which will not be touched)
# local_test_set refers to the test I will be testing for accuracy and other statistics 
training_temp, local_test_set = train_test_split(additional_features_merged, test_size=0.2, random_state=42, stratify=additional_features_merged['class_name'])

x_trained = training_temp.drop(columns=["image_id", "class_name", "image_path", "class_id"])
print(x_trained.columns)
y_trained = training_temp["class_name"]
svm_model = svm.SVC()
svm_model.fit(x_trained, y_trained)

x_test = local_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_test = local_test_set["class_name"]
predictions = svm_model.predict(x_test)
accuracy = accuracy_score(predictions, y_test)
print(accuracy)

```

```python
import torch
from torchvision import models, transforms
import numpy as np
import pandas as pd
import os
from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# ----------------------------
# 1. Load pretrained ResNet50
# ----------------------------
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
model.eval()

feature_extractor = torch.nn.Sequential(*list(model.children())[:-1])


transform_pipeline = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

BASE_DIR = os.getcwd() + "/task1_data/"


def extract_features(image_path):
    image = Image.open(image_path).convert("RGB")
    image = transform_pipeline(image)
    image = image.unsqueeze(0)

    with torch.no_grad():
        features = feature_extractor(image)

    return features.numpy().flatten()

def build_feature_matrix(df):
    features = []
    for path in df["image_path"]:
        full_path = os.path.join(BASE_DIR, path)
        features.append(extract_features(full_path))
    return np.array(features)

X = build_feature_matrix(train_metadata_df)

# ----------------------------
# 5. Merge labels
# ----------------------------
data = pd.merge(train_metadata_df, pd.DataFrame(X), left_index=True, right_index=True)

# ----------------------------
# 6. Train/validation split
# ----------------------------
training_temp, local_test_set = train_test_split(
    data,
    test_size=0.2,
    random_state=42,
    stratify=data["class_id"]
)

x_trained = training_temp.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_trained = training_temp["class_id"]

x_val = local_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_val = local_test_set["class_id"]


scaler = StandardScaler()
x_trained = scaler.fit_transform(x_trained)
x_val = scaler.transform(x_val)


svm_model = svm.SVC(C=10, kernel='rbf', gamma='scale')
svm_model.fit(x_trained, y_trained)

val_pred = svm_model.predict(x_val)
print("Validation Accuracy:", accuracy_score(y_val, val_pred))


def extract_dataset_features(image_paths):
    return scaler.transform([
        extract_features(os.path.join(BASE_DIR, p))
        for p in image_paths
    ])

test_features = extract_dataset_features(test_metadata_df["image_path"])

# 
kaggle_predictions_svm = svm_model.predict(test_features)

submission_df = pd.DataFrame({
    "image_id": test_metadata_df["image_id"],
    "class_id": kaggle_predictions_svm
})

submission_df.to_csv("submission_one.csv", index=False)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

combined_features_merged = pd.merge(train_metadata_df, hog_pca_df, on="image_id", how="inner")
combined_features_merged = pd.merge(combined_features_merged, additional_features_df, on="image_id", how="inner")

combined_training_temp, combined_test_set = train_test_split(
    combined_features_merged,
    test_size=0.2,
    random_state=42,
    stratify=combined_features_merged["class_id"]
)

combined_x_trained = combined_training_temp.drop(
    columns=["image_id", "class_name", "image_path", "class_id"]
)

combined_y_trained = combined_training_temp["class_id"]

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(combined_x_trained)
X_test_scaled = scaler.transform(
    combined_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
)

# Sequential Feature selection

base_model = LogisticRegression(max_iter=1000)

sfs = SequentialFeatureSelector(
    base_model,
    n_features_to_select=50,   
    direction='forward',
    cv=5,
    n_jobs=-1
)

sfs.fit(X_train_scaled, combined_y_trained)

X_train_sel = sfs.transform(X_train_scaled)
X_test_sel = sfs.transform(X_test_scaled)

final_model = LogisticRegression(max_iter=1000)
final_model.fit(X_train_sel, combined_y_trained)

pred = final_model.predict(X_test_sel)

print(f"Accuracy: {accuracy_score(combined_test_set['class_id'], pred)}")
```
