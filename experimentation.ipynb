```python
import torch
from torchvision import models, transforms
import numpy as np
import pandas as pd
import os
from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# ----------------------------
# 1. Load pretrained ResNet50
# ----------------------------
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
model.eval()

feature_extractor = torch.nn.Sequential(*list(model.children())[:-1])


transform_pipeline = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

BASE_DIR = os.getcwd() + "/task1_data/"


def extract_features(image_path):
    image = Image.open(image_path).convert("RGB")
    image = transform_pipeline(image)
    image = image.unsqueeze(0)

    with torch.no_grad():
        features = feature_extractor(image)

    return features.numpy().flatten()

def build_feature_matrix(df):
    features = []
    for path in df["image_path"]:
        full_path = os.path.join(BASE_DIR, path)
        features.append(extract_features(full_path))
    return np.array(features)

X = build_feature_matrix(train_metadata_df)

# ----------------------------
# 5. Merge labels
# ----------------------------
data = pd.merge(train_metadata_df, pd.DataFrame(X), left_index=True, right_index=True)

# ----------------------------
# 6. Train/validation split
# ----------------------------
training_temp, local_test_set = train_test_split(
    data,
    test_size=0.2,
    random_state=42,
    stratify=data["class_id"]
)

x_trained = training_temp.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_trained = training_temp["class_id"]

x_val = local_test_set.drop(columns=["image_id", "class_name", "image_path", "class_id"])
y_val = local_test_set["class_id"]


scaler = StandardScaler()
x_trained = scaler.fit_transform(x_trained)
x_val = scaler.transform(x_val)


svm_model = svm.SVC(C=10, kernel='rbf', gamma='scale')
svm_model.fit(x_trained, y_trained)

val_pred = svm_model.predict(x_val)
print("Validation Accuracy:", accuracy_score(y_val, val_pred))


def extract_dataset_features(image_paths):
    return scaler.transform([
        extract_features(os.path.join(BASE_DIR, p))
        for p in image_paths
    ])

test_features = extract_dataset_features(test_metadata_df["image_path"])

# 
kaggle_predictions_svm = svm_model.predict(test_features)

submission_df = pd.DataFrame({
    "image_id": test_metadata_df["image_id"],
    "class_id": kaggle_predictions_svm
})

submission_df.to_csv("submission_one.csv", index=False)
```
